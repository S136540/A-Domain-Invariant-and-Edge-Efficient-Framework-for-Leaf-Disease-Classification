```python
# ===============================================================
# SimCLR + MADA-lite + SE-ResNet + INT8 Quantization
# 6 Source Classes → RiceDiseases Target
# Fixed: Target class mismatch, per-class metrics, macro F1,
# ROC-AUC, confusion matrix, INT8 accuracy drop
# ===============================================================

import os
import time
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    precision_recall_fscore_support,
    roc_auc_score
)
from sklearn.preprocessing import label_binarize

# ================= CONFIG =================
IMG_SIZE = 48
BATCH_SIZE = 32
EPOCHS_DA = 20
EPOCHS_LINEAR = 30
STEPS_PER_EPOCH = 200

TEMPERATURE = 0.1
FEATURE_DIM = 64
PROJECTION_DIM = 64
NUM_DOMAINS = 2

SOURCE_DIR = '/content/PlantVillage-Dataset/raw/segmented'
TARGET_DIR = '/content/RiceDiseases-DataSet'

SOURCE_CLASSES = [
    "Tomato___Bacterial_spot",
    "Corn_(maize)___Northern_Leaf_Blight",
    "Potato___Early_blight",
    "Corn_(maize)___Cercospora_leaf_spot Gray_leaf_spot",
    "Tomato___Septoria_leaf_spot",
    "Strawberry___Leaf_scorch"
]

# IMPORTANT: manually set target classes to avoid .git or extra folders
# Check folder names using: print(os.listdir(TARGET_DIR))
TARGET_CLASSES = [
    "Bacterial leaf blight",
    "blast",
    "brownspot"
]

# If your folders are different, use this instead:
# TARGET_CLASSES = [
#     "Bacterial leaf blight",
#     "Blast",
#     "Brown spot"
# ]

print("Source Classes:", SOURCE_CLASSES)
print("Target Classes:", TARGET_CLASSES)

# ================= SAFETY CHECK =================
for cls in TARGET_CLASSES:
    path = os.path.join(TARGET_DIR, cls)
    if not os.path.isdir(path):
        raise FileNotFoundError(f"Target class folder not found: {path}")

for cls in SOURCE_CLASSES:
    path = os.path.join(SOURCE_DIR, cls)
    if not os.path.isdir(path):
        raise FileNotFoundError(f"Source class folder not found: {path}")

# ================= GRADIENT REVERSAL =================
@tf.custom_gradient
def grad_reverse(x, lambd):
    def grad(dy):
        return -lambd * dy, None
    return x, grad

class GradientReversal(layers.Layer):
    def __init__(self):
        super().__init__()
        self.lambd = tf.Variable(0.0, trainable=False, dtype=tf.float32)

    def call(self, x):
        return grad_reverse(x, self.lambd)

# ================= AUGMENTATION =================
def strong_aug(x):
    x = tf.image.random_brightness(x, 0.5)
    x = tf.image.random_contrast(x, 0.6, 1.4)
    x = tf.image.random_flip_left_right(x)
    return tf.clip_by_value(x, 0, 1)

def weak_aug(x):
    x = tf.image.random_brightness(x, 0.3)
    x = tf.image.random_contrast(x, 0.8, 1.2)
    x = tf.image.random_flip_left_right(x)
    return tf.clip_by_value(x, 0, 1)

def create_views(img, is_source):
    img = tf.cast(img, tf.float32) / 255.0
    aug = strong_aug if is_source else weak_aug
    return aug(img), aug(img)

# ================= DATA GENERATOR =================
def combined_generator():
    datagen = ImageDataGenerator()

    s_gen = datagen.flow_from_directory(
        SOURCE_DIR,
        classes=SOURCE_CLASSES,
        target_size=(IMG_SIZE, IMG_SIZE),
        batch_size=BATCH_SIZE // 2,
        shuffle=True,
        class_mode=None
    )

    t_gen = datagen.flow_from_directory(
        TARGET_DIR,
        classes=TARGET_CLASSES,
        target_size=(IMG_SIZE, IMG_SIZE),
        batch_size=BATCH_SIZE // 2,
        shuffle=True,
        class_mode=None
    )

    while True:
        xs = next(s_gen)
        xt = next(t_gen)

        X = np.concatenate([xs, xt], axis=0)

        domain = tf.concat([
            tf.zeros(xs.shape[0], dtype=tf.int32),
            tf.ones(xt.shape[0], dtype=tf.int32)
        ], axis=0)

        v1, v2 = [], []

        for i, img in enumerate(X):
            a, b = create_views(img, i < xs.shape[0])
            v1.append(a)
            v2.append(b)

        yield (tf.stack(v1), tf.stack(v2)), domain

# ================= SE-RESNET BACKBONE =================
def se_block(x, ratio=8):
    filters = x.shape[-1]

    se = layers.GlobalAveragePooling2D()(x)
    se = layers.Dense(max(filters // ratio, 8), activation='relu')(se)
    se = layers.Dense(filters, activation='sigmoid')(se)
    se = layers.Reshape((1, 1, filters))(se)

    return layers.multiply([x, se])

def residual_block(x, filters, stride=1):
    shortcut = x

    x = layers.DepthwiseConv2D(
        3,
        strides=stride,
        padding='same',
        use_bias=False
    )(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    x = layers.Conv2D(
        filters,
        1,
        padding='same',
        use_bias=False
    )(x)
    x = layers.BatchNormalization()(x)

    x = se_block(x)

    if stride != 1 or shortcut.shape[-1] != filters:
        shortcut = layers.Conv2D(
            filters,
            1,
            strides=stride,
            padding='same'
        )(shortcut)

    x = layers.Add()([x, shortcut])
    return layers.ReLU()(x)

def build_encoder():
    inp = layers.Input((IMG_SIZE, IMG_SIZE, 3))

    x = layers.Conv2D(16, 3, padding='same')(inp)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    for f, s in zip([32, 32, 64, 64, 128, 128], [1, 1, 2, 1, 2, 1]):
        x = residual_block(x, f, s)

    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(FEATURE_DIM, activation='relu', name='shared_features')(x)

    return models.Model(inp, x, name="encoder_agronet")

def projection_head():
    inp = layers.Input((FEATURE_DIM,))
    x = layers.Dense(32, activation='relu')(inp)
    out = layers.Dense(PROJECTION_DIM)(x)
    return models.Model(inp, out, name="projection")

def domain_head(grl):
    inp = layers.Input((FEATURE_DIM,))
    x = grl(inp)
    x = layers.Dense(32, activation='relu')(x)
    out = layers.Dense(NUM_DOMAINS, activation='softmax')(x)
    return models.Model(inp, out, name="domain")

# ================= NT-XENT LOSS =================
def nt_xent(z1, z2):
    z1 = tf.math.l2_normalize(z1, axis=1)
    z2 = tf.math.l2_normalize(z2, axis=1)

    z = tf.concat([z1, z2], axis=0)

    sim = tf.matmul(z, z, transpose_b=True)
    sim -= tf.eye(tf.shape(z)[0]) * 1e9

    bs = tf.shape(z1)[0]
    labels = tf.concat([
        tf.range(bs, 2 * bs),
        tf.range(bs)
    ], axis=0)

    loss = tf.keras.losses.sparse_categorical_crossentropy(
        labels,
        sim / TEMPERATURE,
        from_logits=True
    )

    return tf.reduce_mean(loss)

# ================= SIMCLR + MADA-LITE MODEL =================
class SimCLR_DA(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.encoder = build_encoder()
        self.projector = projection_head()
        self.grl = GradientReversal()
        self.domain = domain_head(self.grl)

    def train_step(self, data):
        (v1, v2), dom = data

        with tf.GradientTape() as tape:
            h1 = self.encoder(v1, training=True)
            h2 = self.encoder(v2, training=True)

            z1 = self.projector(h1, training=True)
            z2 = self.projector(h2, training=True)

            loss_c = nt_xent(z1, z2)

            h_all = tf.concat([h1, h2], axis=0)
            dom_all = tf.concat([dom, dom], axis=0)

            pseudo = tf.stop_gradient(tf.nn.softmax(h_all))
            weights = tf.reduce_max(pseudo, axis=1)

            dom_pred = self.domain(h_all, training=True)

            loss_d_raw = tf.keras.losses.sparse_categorical_crossentropy(
                dom_all,
                dom_pred
            )

            loss_d = tf.reduce_mean(loss_d_raw * weights)

            loss = loss_c + loss_d

        grads = tape.gradient(loss, self.trainable_variables)
        self.optimizer.apply_gradients(zip(grads, self.trainable_variables))

        return {
            "contrastive_loss": loss_c,
            "domain_loss": loss_d,
            "total_loss": loss
        }

# ================= PHASE 1: UNSUPERVISED TRAINING =================
model = SimCLR_DA()
model.compile(optimizer=tf.keras.optimizers.Adam(1e-4))

gen = combined_generator()

print("\nTraining SimCLR + MADA-lite Domain Adaptation")

for epoch in range(EPOCHS_DA):
    model.grl.lambd.assign(min(1.0, epoch / 10.0))

    print(
        f"\nEpoch {epoch + 1}/{EPOCHS_DA} | "
        f"DOMAIN_LAMBDA = {model.grl.lambd.numpy():.2f}"
    )

    model.fit(
        gen,
        steps_per_epoch=STEPS_PER_EPOCH,
        epochs=1,
        verbose=1
    )

model.encoder.save("encoder_da_agronet.h5")

# ================= PHASE 2: FROZEN ENCODER LINEAR EVALUATION =================
model.encoder.trainable = False

classifier = tf.keras.Sequential([
    layers.Input((IMG_SIZE, IMG_SIZE, 3)),
    model.encoder,
    layers.Dense(len(TARGET_CLASSES), activation='softmax')
])

classifier.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_gen = datagen.flow_from_directory(
    TARGET_DIR,
    classes=TARGET_CLASSES,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    subset='training',
    class_mode='categorical',
    shuffle=True
)

val_gen = datagen.flow_from_directory(
    TARGET_DIR,
    classes=TARGET_CLASSES,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    subset='validation',
    class_mode='categorical',
    shuffle=False
)

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_accuracy",
        patience=7,
        restore_best_weights=True,
        mode="max"
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=3,
        min_lr=1e-6
    )
]

print("\nTraining Linear Classifier on Frozen Encoder")

classifier.fit(
    train_gen,
    epochs=EPOCHS_LINEAR,
    validation_data=val_gen,
    callbacks=callbacks,
    verbose=1
)

loss, acc = classifier.evaluate(val_gen, verbose=0)

# ================= SAVE FP32 MODEL =================
classifier.save("full_da_model_agronet.h5")

encoder_size = os.path.getsize("encoder_da_agronet.h5") / 1024
full_size = os.path.getsize("full_da_model_agronet.h5") / 1024

params = classifier.count_params()
ram_fp32_mb = params * 4 / (1024 ** 2)

print("\n==============================")
print("FP32 MODEL RESULT")
print("==============================")
print(f"Target Accuracy : {acc * 100:.2f}%")
print(f"Encoder Size    : {encoder_size:.2f} KB")
print(f"Full Model Size : {full_size:.2f} KB")
print(f"FP32 RAM Est.   : {ram_fp32_mb:.2f} MB")
print("==============================")

# ================= FULL EVALUATION METRICS =================
val_gen.reset()

y_true = val_gen.classes
y_prob = classifier.predict(val_gen, verbose=1)
y_pred = np.argmax(y_prob, axis=1)

class_names = list(val_gen.class_indices.keys())
num_classes = len(class_names)

present_labels = sorted(list(set(y_true) | set(y_pred)))
present_class_names = [class_names[i] for i in present_labels]

print("\n==============================")
print("TARGET-DOMAIN EVALUATION METRICS")
print("==============================")
print(f"Accuracy: {acc * 100:.2f}%")

print("\nAll Classes:")
print(class_names)

print("\nPresent Labels:")
print(present_labels)

print("\nClassification Report:")
print(
    classification_report(
        y_true,
        y_pred,
        labels=present_labels,
        target_names=present_class_names,
        digits=4,
        zero_division=0
    )
)

macro_precision, macro_recall, macro_f1, _ = precision_recall_fscore_support(
    y_true,
    y_pred,
    labels=np.arange(num_classes),
    average="macro",
    zero_division=0
)

weighted_precision, weighted_recall, weighted_f1, _ = precision_recall_fscore_support(
    y_true,
    y_pred,
    labels=np.arange(num_classes),
    average="weighted",
    zero_division=0
)

print("\nMacro Metrics:")
print(f"Macro Precision : {macro_precision * 100:.2f}%")
print(f"Macro Recall    : {macro_recall * 100:.2f}%")
print(f"Macro F1-score  : {macro_f1 * 100:.2f}%")

print("\nWeighted Metrics:")
print(f"Weighted Precision : {weighted_precision * 100:.2f}%")
print(f"Weighted Recall    : {weighted_recall * 100:.2f}%")
print(f"Weighted F1-score  : {weighted_f1 * 100:.2f}%")

precision, recall, f1, support = precision_recall_fscore_support(
    y_true,
    y_pred,
    labels=np.arange(num_classes),
    zero_division=0
)

per_class_df = pd.DataFrame({
    "Class": class_names,
    "Precision (%)": precision * 100,
    "Recall (%)": recall * 100,
    "F1-score (%)": f1 * 100,
    "Support": support
})

print("\nPer-Class Metrics:")
print(per_class_df)

per_class_df.to_csv("per_class_metrics.csv", index=False)

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(num_classes)
)

plt.figure(figsize=(8, 6))
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)
disp.plot(cmap="Blues", values_format="d", xticks_rotation=45)
plt.title("Confusion Matrix - Cross-Domain Evaluation")
plt.tight_layout()
plt.savefig("confusion_matrix_cross_domain.png", dpi=300)
plt.show()

try:
    y_true_bin = label_binarize(
        y_true,
        classes=np.arange(num_classes)
    )

    macro_auc = roc_auc_score(
        y_true_bin,
        y_prob,
        average="macro",
        multi_class="ovr"
    )

    weighted_auc = roc_auc_score(
        y_true_bin,
        y_prob,
        average="weighted",
        multi_class="ovr"
    )

    print("\nROC-AUC Metrics:")
    print(f"Macro ROC-AUC    : {macro_auc:.4f}")
    print(f"Weighted ROC-AUC : {weighted_auc:.4f}")

except Exception as e:
    macro_auc = None
    weighted_auc = None
    print("\nROC-AUC could not be calculated.")
    print("Reason:", e)

summary_df = pd.DataFrame({
    "Metric": [
        "Accuracy (%)",
        "Macro Precision (%)",
        "Macro Recall (%)",
        "Macro F1-score (%)",
        "Weighted Precision (%)",
        "Weighted Recall (%)",
        "Weighted F1-score (%)",
        "Macro ROC-AUC",
        "Weighted ROC-AUC"
    ],
    "Value": [
        acc * 100,
        macro_precision * 100,
        macro_recall * 100,
        macro_f1 * 100,
        weighted_precision * 100,
        weighted_recall * 100,
        weighted_f1 * 100,
        macro_auc,
        weighted_auc
    ]
})

print("\nSummary Metrics:")
print(summary_df)

summary_df.to_csv("summary_evaluation_metrics.csv", index=False)

print("\nSaved:")
print("per_class_metrics.csv")
print("summary_evaluation_metrics.csv")
print("confusion_matrix_cross_domain.png")

# ================= TFLITE CONVERSION + PTQ INT8 QUANTIZATION =================
def representative_data_gen():
    train_gen.reset()

    for images, _ in train_gen:
        yield [images.astype(np.float32)]
        break

tflite_model_file = "full_da_model_agronet_int8_small.tflite"

converter = tf.lite.TFLiteConverter.from_keras_model(classifier)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = representative_data_gen
converter.target_spec.supported_ops = [
    tf.lite.OpsSet.TFLITE_BUILTINS_INT8
]
converter.inference_input_type = tf.uint8
converter.inference_output_type = tf.uint8

tflite_model = converter.convert()

with open(tflite_model_file, "wb") as f:
    f.write(tflite_model)

tflite_size = os.path.getsize(tflite_model_file) / 1024

print("\n==============================")
print("PTQ INT8 MODEL SIZE")
print("==============================")
print(f"INT8 TFLite Model Size: {tflite_size:.2f} KB")

# ================= INT8 TFLITE ACCURACY EVALUATION =================
def evaluate_tflite_model(tflite_path, val_gen):
    interpreter = tf.lite.Interpreter(model_path=tflite_path)
    interpreter.allocate_tensors()

    input_details = interpreter.get_input_details()
    output_details = interpreter.get_output_details()

    input_scale, input_zero_point = input_details[0]["quantization"]
    output_scale, output_zero_point = output_details[0]["quantization"]

    y_true_int8 = []
    y_pred_int8 = []

    val_gen.reset()
    total_seen = 0

    for batch_images, batch_labels in val_gen:
        for i in range(batch_images.shape[0]):
            img = batch_images[i:i+1].astype(np.float32)

            if input_details[0]["dtype"] == np.uint8:
                img = img / input_scale + input_zero_point
                img = np.clip(img, 0, 255).astype(np.uint8)

            interpreter.set_tensor(
                input_details[0]["index"],
                img
            )
            interpreter.invoke()

            output = interpreter.get_tensor(
                output_details[0]["index"]
            )

            if output_details[0]["dtype"] == np.uint8:
                output = (
                    output.astype(np.float32) - output_zero_point
                ) * output_scale

            pred = np.argmax(output, axis=1)[0]
            true = np.argmax(batch_labels[i])

            y_pred_int8.append(pred)
            y_true_int8.append(true)

            total_seen += 1

            if total_seen >= val_gen.samples:
                break

        if total_seen >= val_gen.samples:
            break

    y_true_int8 = np.array(y_true_int8)
    y_pred_int8 = np.array(y_pred_int8)

    int8_acc = np.mean(y_true_int8 == y_pred_int8)

    return int8_acc, y_true_int8, y_pred_int8

int8_acc, y_true_int8, y_pred_int8 = evaluate_tflite_model(
    tflite_model_file,
    val_gen
)

accuracy_drop = (acc - int8_acc) * 100

print("\n==============================")
print("PTQ INT8 QUANTIZATION RESULT")
print("==============================")
print(f"FP32 Accuracy     : {acc * 100:.2f}%")
print(f"INT8 Accuracy     : {int8_acc * 100:.2f}%")
print(f"Accuracy Drop     : {accuracy_drop:.2f}%")
print(f"INT8 Model Size   : {tflite_size:.2f} KB")
print("==============================")

quant_df = pd.DataFrame({
    "Model": ["SimCLR + MADA-Lite"],
    "Quantization Type": ["Post-Training INT8 Quantization"],
    "FP32 Accuracy (%)": [acc * 100],
    "INT8 Accuracy (%)": [int8_acc * 100],
    "Accuracy Drop (%)": [accuracy_drop],
    "Encoder Size (KB)": [encoder_size],
    "FP32 Model Size (KB)": [full_size],
    "INT8 Model Size (KB)": [tflite_size],
    "FP32 RAM Estimate (MB)": [ram_fp32_mb]
})

print("\nQuantization Summary:")
print(quant_df)

quant_df.to_csv("quantization_summary.csv", index=False)

# ================= FINAL CONSOLIDATED REPORT =================
print("\n" + "=" * 60)
print("FINAL CONSOLIDATED REPORT")
print("=" * 60)
print(f"Target Accuracy       : {acc * 100:.2f}%")
print(f"Macro Precision       : {macro_precision * 100:.2f}%")
print(f"Macro Recall          : {macro_recall * 100:.2f}%")
print(f"Macro F1-score        : {macro_f1 * 100:.2f}%")
print(f"Macro ROC-AUC         : {macro_auc}")
print(f"Weighted ROC-AUC      : {weighted_auc}")
print(f"Encoder Size          : {encoder_size:.2f} KB")
print(f"FP32 Model Size       : {full_size:.2f} KB")
print(f"INT8 Model Size       : {tflite_size:.2f} KB")
print(f"INT8 Accuracy         : {int8_acc * 100:.2f}%")
print(f"Accuracy Drop         : {accuracy_drop:.2f}%")
print(f"FP32 RAM Estimate     : {ram_fp32_mb:.2f} MB")
print("=" * 60)
print("Saved files:")
print("1. encoder_da_agronet.h5")
print("2. full_da_model_agronet.h5")
print("3. full_da_model_agronet_int8_small.tflite")
print("4. per_class_metrics.csv")
print("5. summary_evaluation_metrics.csv")
print("6. quantization_summary.csv")
print("7. confusion_matrix_cross_domain.png")
print("=" * 60)
```

```python
print("\nMacro Metrics:")
print(f"Macro Precision : {macro_precision * 100:.2f}%")
print(f"Macro Recall    : {macro_recall * 100:.2f}%")
print(f"Macro F1-score  : {macro_f1 * 100:.2f}%")

print("\nWeighted Metrics:")
print(f"Weighted Precision : {weighted_precision * 100:.2f}%")
print(f"Weighted Recall    : {weighted_recall * 100:.2f}%")
print(f"Weighted F1-score  : {weighted_f1 * 100:.2f}%")

precision, recall, f1, support = precision_recall_fscore_support(
    y_true,
    y_pred,
    labels=np.arange(num_classes),
    zero_division=0
)

per_class_df = pd.DataFrame({
    "Class": class_names,
    "Precision (%)": precision * 100,
    "Recall (%)": recall * 100,
    "F1-score (%)": f1 * 100,
    "Support": support
})

print("\nPer-Class Metrics:")
print(per_class_df)

per_class_df.to_csv("per_class_metrics.csv", index=False)

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(num_classes)
)

plt.figure(figsize=(8, 6))
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)
disp.plot(cmap="YlOrBr")
plt.title("Confusion Matrix - Cross-Domain Evaluation")
plt.tight_layout()
plt.savefig("confusion_matrix_cross_domain.png", dpi=300)
plt.show()

try:
    y_true_bin = label_binarize(
        y_true,
        classes=np.arange(num_classes)
    )

    macro_auc = roc_auc_score(
        y_true_bin,
        y_prob,
        average="macro",
        multi_class="ovr"
    )

    weighted_auc = roc_auc_score(
        y_true_bin,
        y_prob,
        average="weighted",
        multi_class="ovr"
    )

    print("\nROC-AUC Metrics:")
    print(f"Macro ROC-AUC    : {macro_auc:.4f}")
    print(f"Weighted ROC-AUC : {weighted_auc:.4f}")

except Exception as e:
    macro_auc = None
    weighted_auc = None
    print("\nROC-AUC could not be calculated.")
    print("Reason:", e)

summary_df = pd.DataFrame({
    "Metric": [
        "Accuracy (%)",
        "Macro Precision (%)",
        "Macro Recall (%)",
        "Macro F1-score (%)",
        "Weighted Precision (%)",
        "Weighted Recall (%)",
        "Weighted F1-score (%)",
        "Macro ROC-AUC",
        "Weighted ROC-AUC"
    ],
    "Value": [
        acc * 100,
        macro_precision * 100,
        macro_recall * 100,
        macro_f1 * 100,
        weighted_precision * 100,
        weighted_recall * 100,
        weighted_f1 * 100,
        macro_auc,
        weighted_auc
    ]
})

print("\nSummary Metrics:")
print(summary_df)

summary_df.to_csv("summary_evaluation_metrics.csv", index=False)

print("\nSaved:")
print("per_class_metrics.csv")
print("summary_evaluation_metrics.csv")
print("confusion_matrix_cross_domain.png")

# ================= TFLITE CONVERSION + PTQ INT8 QUANTIZATION =================
def representative_data_gen():
    train_gen.reset()

    for images, _ in train_gen:
        yield [images.astype(np.float32)]
        break

tflite_model_file = "full_da_model_agronet_int8_small.tflite"

converter = tf.lite.TFLiteConverter.from_keras_model(classifier)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = representative_data_gen
converter.target_spec.supported_ops = [
    tf.lite.OpsSet.TFLITE_BUILTINS_INT8
]
converter.inference_input_type = tf.uint8
converter.inference_output_type = tf.uint8

tflite_model = converter.convert()

with open(tflite_model_file, "wb") as f:
    f.write(tflite_model)

tflite_size = os.path.getsize(tflite_model_file) / 1024

print("\n==============================")
print("PTQ INT8 MODEL SIZE")
print("==============================")
print(f"INT8 TFLite Model Size: {tflite_size:.2f} KB")

# ================= INT8 TFLITE ACCURACY EVALUATION =================
def evaluate_tflite_model(tflite_path, val_gen):
    interpreter = tf.lite.Interpreter(model_path=tflite_path)
    interpreter.allocate_tensors()

    input_details = interpreter.get_input_details()
    output_details = interpreter.get_output_details()

    input_scale, input_zero_point = input_details[0]["quantization"]
    output_scale, output_zero_point = output_details[0]["quantization"]

    y_true_int8 = []
    y_pred_int8 = []

    val_gen.reset()
    total_seen = 0

    for batch_images, batch_labels in val_gen:
        for i in range(batch_images.shape[0]):
            img = batch_images[i:i+1].astype(np.float32)

            if input_details[0]["dtype"] == np.uint8:
                img = img / input_scale + input_zero_point
                img = np.clip(img, 0, 255).astype(np.uint8)

            interpreter.set_tensor(
                input_details[0]["index"],
                img
            )
            interpreter.invoke()

            output = interpreter.get_tensor(
                output_details[0]["index"]
            )

            if output_details[0]["dtype"] == np.uint8:
                output = (
                    output.astype(np.float32) - output_zero_point
                ) * output_scale

            pred = np.argmax(output, axis=1)[0]
            true = np.argmax(batch_labels[i])

            y_pred_int8.append(pred)
            y_true_int8.append(true)

            total_seen += 1

            if total_seen >= val_gen.samples:
                break

        if total_seen >= val_gen.samples:
            break

    y_true_int8 = np.array(y_true_int8)
    y_pred_int8 = np.array(y_pred_int8)

    int8_acc = np.mean(y_true_int8 == y_pred_int8)

    return int8_acc, y_true_int8, y_pred_int8

int8_acc, y_true_int8, y_pred_int8 = evaluate_tflite_model(
    tflite_model_file,
    val_gen
)

accuracy_drop = (acc - int8_acc) * 100

print("\n==============================")
print("PTQ INT8 QUANTIZATION RESULT")
print("==============================")
print(f"FP32 Accuracy     : {acc * 100:.2f}%")
print(f"INT8 Accuracy     : {int8_acc * 100:.2f}%")
print(f"Accuracy Drop     : {accuracy_drop:.2f}%")
print(f"INT8 Model Size   : {tflite_size:.2f} KB")
print("==============================")

quant_df = pd.DataFrame({
    "Model": ["SimCLR + MADA-Lite"],
    "Quantization Type": ["Post-Training INT8 Quantization"],
    "FP32 Accuracy (%)": [acc * 100],
    "INT8 Accuracy (%)": [int8_acc * 100],
    "Accuracy Drop (%)": [accuracy_drop],
    "Encoder Size (KB)": [encoder_size],
    "FP32 Model Size (KB)": [full_size],
    "INT8 Model Size (KB)": [tflite_size],
    "FP32 RAM Estimate (MB)": [ram_fp32_mb]
})

print("\nQuantization Summary:")
print(quant_df)

quant_df.to_csv("quantization_summary.csv", index=False)

# ================= FINAL CONSOLIDATED REPORT =================
print("\n" + "=" * 60)
print("FINAL CONSOLIDATED REPORT")
print("=" * 60)
print(f"Target Accuracy       : {acc * 100:.2f}%")
print(f"Macro Precision       : {macro_precision * 100:.2f}%")
print(f"Macro Recall          : {macro_recall * 100:.2f}%")
print(f"Macro F1-score        : {macro_f1 * 100:.2f}%")
print(f"Macro ROC-AUC         : {macro_auc}")
print(f"Weighted ROC-AUC      : {weighted_auc}")
print(f"Encoder Size          : {encoder_size:.2f} KB")
print(f"FP32 Model Size       : {full_size:.2f} KB")
print(f"INT8 Model Size       : {tflite_size:.2f} KB")
print(f"INT8 Accuracy         : {int8_acc * 100:.2f}%")
print(f"Accuracy Drop         : {accuracy_drop:.2f}%")
print(f"FP32 RAM Estimate     : {ram_fp32_mb:.2f} MB")
print("=" * 60)
print("Saved files:")
print("1. encoder_da_agronet.h5")
print("2. full_da_model_agronet.h5")
print("3. full_da_model_agronet_int8_small.tflite")
print("4. per_class_metrics.csv")
print("5. summary_evaluation_metrics.csv")
print("6. quantization_summary.csv")
print("7. confusion_matrix_cross_domain.png")
print("=" * 60)
```

```python
# ===============================================================
# Accuracy-Loss Curve Code
# Uses Saved Frozen Encoder: encoder_da_agronet.h5
# Trains only Linear Classifier Again
# ===============================================================

import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# ================= CONFIG =================
IMG_SIZE = 48
BATCH_SIZE = 32
EPOCHS_LINEAR = 20

TARGET_DIR = '/content/RiceDiseases-DataSet'

TARGET_CLASSES = [
    "Bacterial leaf blight",
    "blast",
    "brownspot"
]

ENCODER_PATH = "encoder_da_agronet.h5"

# ================= CHECKS =================
if not os.path.exists(ENCODER_PATH):
    raise FileNotFoundError(
        f"{ENCODER_PATH} not found. Please make sure the trained encoder is saved."
    )

for cls in TARGET_CLASSES:
    path = os.path.join(TARGET_DIR, cls)
    if not os.path.isdir(path):
        raise FileNotFoundError(f"Target class folder not found: {path}")

# ================= LOAD FROZEN ENCODER =================
encoder = tf.keras.models.load_model(
    ENCODER_PATH,
    compile=False
)

encoder.trainable = False

# ================= DATA GENERATORS =================
datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_gen = datagen.flow_from_directory(
    TARGET_DIR,
    classes=TARGET_CLASSES,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    subset='training',
    class_mode='categorical',
    shuffle=True
)

val_gen = datagen.flow_from_directory(
    TARGET_DIR,
    classes=TARGET_CLASSES,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    subset='validation',
    class_mode='categorical',
    shuffle=False
)

# ================= LINEAR CLASSIFIER =================
classifier = tf.keras.Sequential([
    layers.Input((IMG_SIZE, IMG_SIZE, 3)),
    encoder,
    layers.Dense(len(TARGET_CLASSES), activation='softmax')
])

classifier.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_accuracy",
        patience=7,
        restore_best_weights=True,
        mode="max"
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=3,
        min_lr=1e-6
    )
]

# ================= TRAIN CLASSIFIER AND SAVE HISTORY =================
history = classifier.fit(
    train_gen,
    epochs=EPOCHS_LINEAR,
    validation_data=val_gen,
    callbacks=callbacks,
    verbose=1
)

# ================= EVALUATION =================
loss, acc = classifier.evaluate(val_gen, verbose=0)

print("\n==============================")
print("LINEAR CLASSIFIER RESULT")
print("==============================")
print(f"Validation Accuracy: {acc * 100:.2f}%")
print(f"Validation Loss    : {loss:.4f}")
print("==============================")

# ================= SAVE MODEL =================
classifier.save("linear_probe_with_history.h5")

# ================= ACCURACY CURVE =================
plt.figure(figsize=(7, 5))
plt.plot(
    history.history["accuracy"],
    label="Training Accuracy",
    marker="o"
)
plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy",
    marker="o"
)
plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("training_validation_accuracy_curve.png", dpi=300)
plt.show()

# ================= LOSS CURVE =================
plt.figure(figsize=(7, 5))
plt.plot(
    history.history["loss"],
    label="Training Loss",
    marker="o"
)
plt.plot(
    history.history["val_loss"],
    label="Validation Loss",
    marker="o"
)
plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("training_validation_loss_curve.png", dpi=300)
plt.show()

# ================= SAVE HISTORY AS CSV =================
import pandas as pd

history_df = pd.DataFrame(history.history)
history_df.to_csv("linear_probe_training_history.csv", index=False)

print("\nSaved files:")
print("1. linear_probe_with_history.h5")
print("2. training_validation_accuracy_curve.png")
print("3. training_validation_loss_curve.png")
print("4. linear_probe_training_history.csv")
```
